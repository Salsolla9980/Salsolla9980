```python
#es. pag 8
def fattori(n):
    i = 2
    fattori = []
    while i * i <= n:
        if n % i:
            i += 1
        else:
            n //= i
            fattori.append(i)
    if n > 1:
        fattori.append(n)
    return fattori

n = int(input("Inserisci un numero: "))
print(fattori(n))
```

```python
#es. pag 11
#Versione con ciclo while

N = int(input("Inserisci il numero di potenze: "))
K = int(input("Inserisci la base: "))
potenze = []
i = 0
while i < N:
    potenza = K ** i
    potenze.append(potenza)
    i += 1
print(potenze)


#Versione con ciclo for

N = int(input("Inserisci il numero di potenze: "))
K = int(input("Inserisci la base: "))
potenze = []
for i in range(N):
    potenza = K ** i
    potenze.append(potenza)
print(potenze)
```
